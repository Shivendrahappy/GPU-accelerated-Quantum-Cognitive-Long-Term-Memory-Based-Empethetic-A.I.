Context-Dependent Retrieval Experiment

Testing whether adding emotional context to a query changes which
memory gets retrieved as most relevant. 3 new memories (short
sessions / overwhelm / progress goal), same base question asked
under two contexts - exhausted vs energetic. If top match differs
between the two, that's evidence retrieval is context-sensitive,
not just keyword/topic-sensitive. This is a first honest test of
the paper's context-dependence claim, using only existing tools
(no new quantum math needed).

Superposition-inspired blended query experiment (13 Oct)

Tests a genuine blended state - combines the "exhausted" and
"energetic" context vectors into ONE vector (0.5 weight each)
before retrieval. This is a linear combination, the real structural
operation behind superposition, though without complex phase/
interference - so calling this "superposition-inspired," not
literal quantum superposition.

Comparing this blended result against the two pure-context results
shows whether combining emotional states produces retrieval
behavior different from either pure state alone.

```python
# Context memories for this experiment
context_memories = [
    {"content": "I prefer short study sessions."},
    {"content": "I feel overwhelmed when I study for too long."},
    {"content": "I want to make progress on my research project."}
]

context_contents = [m['content'] for m in context_memories]
context_embeddings = model.encode(context_contents)

base_query = "How should I plan today's study session?"

context_1 = "I feel exhausted. " + base_query
context_2 = "I feel energetic and want to make progress. " + base_query

print("CONTEXT 1 (exhausted):")
results_1 = retrieve(context_1, context_memories, context_embeddings)
for score, mem in results_1:
    print(f"  {score:.3f} | {mem['content']}")

print("\nCONTEXT 2 (energetic):")
results_2 = retrieve(context_2, context_memories, context_embeddings)
for score, mem in results_2:
    print(f"  {score:.3f} | {mem['content']}")

# Blended / superposition-inspired test
exhausted_vec = model.encode(["I feel exhausted"])[0]
energetic_vec = model.encode(["I feel energetic and want to make progress"])[0]
base_vec = model.encode([base_query])[0]

blended_vec = 0.5 * exhausted_vec + 0.5 * energetic_vec + base_vec
blended_vec = blended_vec / np.linalg.norm(blended_vec)

def retrieve_from_vector(query_vec, memory_list, embeddings_list, top_k=3):
    similarities = []
    for i, memory_embedding in enumerate(embeddings_list):
        score = cosine_similarity(query_vec, memory_embedding)
        similarities.append((score, memory_list[i]))
    similarities.sort(key=lambda x: x[0], reverse=True)
    return similarities[:top_k]

print("\nBLENDED (both emotions at once):")
results_blend = retrieve_from_vector(blended_vec, context_memories, context_embeddings)
for score, mem in results_blend:
    print(f"  {score:.3f} | {mem['content']}")
```

Stronger context variant (13 Oct) - testing if more extreme/distinct
emotional framing produces context-sensitive ranking changes, since
mild framing ("I feel exhausted") didn't shift the top result.

```python
context_1_strong = "I am completely drained and can barely focus, everything feels like too much right now. " + base_query
context_2_strong = "I am full of energy and motivation, ready to push hard and get a lot done. " + base_query

print("CONTEXT 1 STRONG (exhausted):")
results_1s = retrieve(context_1_strong, context_memories, context_embeddings)
for score, mem in results_1s:
    print(f"  {score:.3f} | {mem['content']}")

print("\nCONTEXT 2 STRONG (energetic):")
results_2s = retrieve(context_2_strong, context_memories, context_embeddings)
for score, mem in results_2s:
    print(f"  {score:.3f} | {mem['content']}")
```

Result: Order-Effect Experiment (13 Oct)

Question tested: Does projecting a query through two context
subspaces in a different order change the final result?

Method: Built two projection matrices from existing memory
embeddings - Context A (Python + TOEFL memories) and Context B
(TOEFL + Writing memories), using QR decomposition to create
valid orthonormal projectors. Projected a test query ("How should
I improve my learning?") through Context A then B, and separately
through Context B then A, renormalizing the vector after each
projection step.

Result:
Context A then B: 0.745
Context B then A: 0.819
Difference: 0.073

Interpretation: This is a genuine order effect - the two results
are different even though the same query and same two contexts
were used, only the sequence changed. Standard cosine similarity
retrieval cannot produce this, since cosine similarity is
symmetric by mathematical definition (similarity(A,B) always
equals similarity(B,A)). This confirms that projection-based
retrieval can mathematically exhibit order-dependence, which is
the same underlying mechanism quantum cognition models use to
explain human question-order effects.

Limitation: This demonstrates mathematical possibility only. It
has not been validated against real human memory-retrieval
behavior, and the two context groupings used here were arbitrary
rather than drawn from a real use case. Future work should test
this with meaningfully defined contexts and compare against actual
user behavior.

```python

# EXPERIMENT 3: Projection order effect

import numpy as np

def build_projector(vectors):
    # Each row is a memory embedding
    Q, _ = np.linalg.qr(vectors.T)
    return Q @ Q.T

def project_and_measure(query_vec, P1, P2):
    step1 = P1 @ query_vec
    norm1 = np.linalg.norm(step1)

    if norm1 < 1e-12:
        return 0.0

    step1 = step1 / norm1
    step2 = P2 @ step1

    return float(np.dot(step2, step2))

# Context A: memories 1 and 2
context_A_vectors = np.asarray([embeddings[0], embeddings[1]])

# Context B: memories 2 and 3
context_B_vectors = np.asarray([embeddings[1], embeddings[2]])

P_A = build_projector(context_A_vectors)
P_B = build_projector(context_B_vectors)

query_vec = model.encode(
    ["How should I improve my learning?"]
)[0]
query_vec = query_vec / np.linalg.norm(query_vec)

result_AB = project_and_measure(query_vec, P_A, P_B)
result_BA = project_and_measure(query_vec, P_B, P_A)

print(f"A then B: {result_AB:.6f}")
print(f"B then A: {result_BA:.6f}")
print(f"Absolute difference: {abs(result_AB - result_BA):.6f}")
print(
    "Numerically different:",
    not np.isclose(result_AB, result_BA, atol=1e-8)
)
```

The experiment produced several important observations:

1. Relevant queries generally received higher similarity scores than unrelated queries.

## 8. Limitations

Several limitations must be considered:

- The memory collection contains only eight manually constructed memories.
- The number of queries is small.
- The queries were manually designed rather than collected from real-world conversations.
- Relevant and unrelated categories were predefined by the experimenter.
- The initial threshold was derived from the initial dataset rather than an independent training set.
- The borderline evaluation is preliminary and is not a substitute for testing on a larger, independently collected dataset.
- Cosine similarity measures vector alignment; it does not represent a probability or a guarantee of semantic relevance.
- The experiment evaluates retrieval relevance, not emotional understanding or the quality of the final response.

Consequently, the observed 100% accuracy should be interpreted only within this experimental dataset.

## 9. Conclusion

Experiment 2E provides preliminary evidence that cosine similarity can support relevance-aware memory retrieval.

The initial evaluation showed a substantial difference between the similarity scores of clearly relevant and unrelated queries. An exploratory threshold of 0.365 was subsequently tested on eight borderline queries, and all eight received the expected classification.

The findings suggest that a threshold-based rejection mechanism may improve a personal-memory system by preventing weakly related memories from automatically entering the response-generation context.

This provides a potential foundation for a more selective retrieval architecture in Fizzy.

However, the threshold requires further evaluation using a larger and independently constructed dataset before it can be considered reliable.

**Overall finding:** Semantic similarity can be used not only to rank candidate memories but also as a preliminary signal for deciding whether a retrieved memory should be accepted or rejected.
