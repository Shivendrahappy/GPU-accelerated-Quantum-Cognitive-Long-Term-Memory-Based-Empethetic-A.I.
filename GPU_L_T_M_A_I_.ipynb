### Retrieving only the content of all memories

To ensure we print the 'content' for each memory, we need to loop through the `loaded_memories` list and explicitly access the `['content']` key for each `memory` dictionary within the loop.

### Retrieving only the content of 'TOEFL study' memories

To ensure we print only the 'content' for memories related to 'TOEFL study', we need to loop through the `loaded_memories` and apply a conditional check on the `context` field, then specifically print the `['content']`.

```python
for memory in loaded_memories:
  if memory['context'] == 'TOEFL study':
    print(memory['content'])
```
