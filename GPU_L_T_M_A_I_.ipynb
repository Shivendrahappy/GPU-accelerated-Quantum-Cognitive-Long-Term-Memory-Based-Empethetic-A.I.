For long term memory we start with making dictionary 'create" as step 1 as memory 1

```python
create = {
    'content': 'I prefer learning Python with small examples',
    'timestamp': '10 Sep 2026',
    'topic': 'learning',
    'importance': 0.8,
    'context': 'python study'
}

print(create)
```
